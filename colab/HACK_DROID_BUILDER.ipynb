# HACK DROID BUILDER - One Click APK Build

Run the cells from top to bottom. The final cell downloads the generated APK.

```python
import os
import shutil
import subprocess
from pathlib import Path

REPO_URL = input("GitHub repository URL: ").strip()
REPO_DIR = Path("/content/hack-droid-builder")
if REPO_DIR.exists():
    shutil.rmtree(REPO_DIR)
subprocess.run(["git", "clone", REPO_URL, str(REPO_DIR)], check=True)
os.chdir(REPO_DIR)
shutil.copy("resources/main.py", "main.py")
shutil.copy("resources/buildozer.spec", "buildozer.spec")
print(f"Source downloaded to {REPO_DIR}")
```
